```python

print("===== GridSense: IoT Smart Grid System =====")

# User Input
house_name = input("Enter House Name: ")
units = int(input("Enter Electricity Units Consumed Today: "))

print("\n----- SMART GRID REPORT -----")
print("House Name:", house_name)
print("Units Consumed:", units)

# Smart Grid Analysis
if units <= 5:
    print("Status: Normal Usage")
    print("Suggestion: Great! Keep saving electricity.")

elif units <= 10:
    print("Status: Moderate Usage")
    print("Suggestion: Turn off unused lights and fans.")

elif units <= 20:
    print("Status: High Usage")
    print("Alert: Electricity consumption is high!")
    print("Suggestion: Check heavy appliances.")

else:
    print("Status: Critical Usage")
    print("ALERT: Possible power wastage detected!")
    print("Suggestion: Contact electricity provider.")

print("\nThank you for using GridSense!")
```
